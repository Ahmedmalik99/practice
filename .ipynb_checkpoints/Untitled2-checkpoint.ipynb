```python
# Write a function called max( ), that returns the maxium of three integer numbers. 

def find_max_of_three(a, b, c):
    return max(a, b, c)  

# Example usage
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))
num3 = int(input("Enter the third number: "))

result = find_max_of_three(num1, num2, num3)
print("The maximum of the three numbers is:", result)
    
```

```python
vehicles = [
    ["Toyota", "ABC123", 2018],
    ["Ford", "XYZ456", 2020],
    ["Honda", "LMN789", 2019],
    ["BMW", "PQR234", 2021],
    ["Tesla", "DEF567", 2022]
]

vehicles_copy = [
    ["Toyota", "NEW123", 2018],
    ["Ford", "XYZ456", 2020],
    ["Honda", "NEW456", 2019],
    ["BMW", "PQR234", 2021],
    ["Tesla", "DEF567", 2022]
]

common_vehicles = []

for vehicle in vehicles:
    for vehicle_copy in vehicles_copy:
        if vehicle[0] == vehicle_copy[0] and vehicle[2] == vehicle_copy[2]:
            common_vehicles.append(vehicle)

print("Common vehicles:")
for vehicle in common_vehicles:
    print(f"Company: {vehicle[0]}, Registration Number: {vehicle[1]}, Manufacturing Year: {vehicle[2]}")
```

```python
vehicles = [
    ["Toyota", "ABC123", 2018],
    ["Ford", "XYZ456", 2020],
    ["Honda", "LMN789", 2019],
    ["BMW", "PQR234", 2021],
    ["Tesla", "DEF567", 2022]
]

vehicles_copy = [
    ["Toyota", "ABC123", 2018],
    ["Ford", "XYZ456", 2020],
    ["Honda", "LMN789", 2019],
    ["BMW", "PQR234", 2021],
    ["Tesla", "DEF567", 2022]
]

vehicles_copy[0][1] = "NEW123"  
vehicles_copy[2][1] = "NEW456"  

print("Original list:")
for vehicle in vehicles:
    print(f"Company: {vehicle[0]}, Registration Number: {vehicle[1]}, Manufacturing Year: {vehicle[2]}")

print("\nCopied list with updated registration numbers:")
for vehicle in vehicles_copy:
    print(f"Company: {vehicle[0]}, Registration Number: {vehicle[1]}, Manufacturing Year: {vehicle[2]}")
```
