```python
import math

def area_circle(radius):
    return math.pi * radius ** 2

def area_square(side):
    return side ** 2

def area_rectangle(length, breadth):
    return length * breadth

def area_cube(side):
    return 6 * side ** 2

def area_triangle(base, height):
    return (base * height) / 2

print("Select a shape to calculate its area:")
print("1. Circle")
print("2. Square")
print("3. Rectangle")
print("4. Cube")
print("5. Triangle")

choice = int(input("Enter your choice (1-5): "))

if choice == 1:  
    radius = float(input("Enter the radius of the circle: "))
    area = area_circle(radius)
    print(f"The area of the circle is: {area:.2f}")

elif choice == 2:  
    side = float(input("Enter the side length of the square: "))
    area = area_square(side)
    print(f"The area of the square is: {area:.2f}")

elif choice == 3:  
    length = float(input("Enter the length of the rectangle: "))
    breadth = float(input("Enter the breadth of the rectangle: "))
    area = area_rectangle(length, breadth)
    print(f"The area of the rectangle is: {area:.2f}")

elif choice == 4:  
    side = float(input("Enter the side length of the cube: "))
    area = area_cube(side)
    print(f"The surface area of the cube is: {area:.2f}")

elif choice == 5:  
    base = float(input("Enter the base length of the triangle: "))
    height = float(input("Enter the height of the triangle: "))
    area = area_triangle(base, height)
    print(f"The area of the triangle is: {area:.2f}")

else:
    print("Invalid choice! Please select a number between 1 and 5.")
```

```python
class Matrix:
    def __init__(self):
        self.data = [[2,4],[6,8]]
        
    def display(self):
        for i in self.data:
            print(i)            
ob1 = Matrix()
print(ob1.display())
```

```python
class Box:
    def __init__(self,l,w,h):
        self.lenth = l
        self.width = w
        self.height = h
        
    def volume(self):
        return self.lenth * self.width * self.height
    
ob = Box(3,4,5)
ob2 = Box(7,3,9)
print(ob.volume())
print(ob2.volume())
```

```python
class Circle:
    def __init__(self,radius):
        self.radius = radius
        
    def area(self):
        return 3.142 * (self.radius**2)
    
ob = Circle(5)
ob2=Circle(3)
print(ob.area())
print(ob2.area())
```

```python
class Employee:
    def __init__(self, name, emp_id, designation, salary):
        self.name = name  
        self.emp_id = emp_id
        self.designation = designation
        self.salary = salary

    def save_to_file(self):
        with open("employee.txt", "a") as file:
            file.write(f"{self.name},{self.emp_id},{self.designation},{self.salary}\n")

    def read_from_file():
        with open("employee.txt", "r") as file:
            for line in file:
                name, emp_id, designation, salary = line.strip().split(",")
                print(f"Name: {name}, ID: {emp_id}, Designation: {designation}, Salary: ${salary}")

emp1 = Employee("Ahmed", 55, "Manager", 50000)
emp1.save_to_file()

print("Employee Records:")
Employee.read_from_file()
```
