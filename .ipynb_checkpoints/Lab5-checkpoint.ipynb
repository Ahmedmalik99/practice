```python
# •	To print the odd number from 10 to 100 and even numbers from 200 to 96.



for i in range(96, 200):
    if i % 2 == 0:
        print(i)

print()

for w in range(10, 101):
    if w % 3 == 0:
        print(w)
```

```python
# •	To take a integer value from a user and print the factorial of that number.

number = int(input("Enter an integer: "))

factorial = 1

for i in range(1, number + 1):
    factorial *= i

print(f"Factorial:", factorial)

```

```python
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

print()
print("The highest value is:", max(num1, num2))
print("The lowest value is:", min(num1, num2))
```

```python
for j in range(3):
    number = int(input(f"Enter integer {j + 1}: "))
    
    factorial = 1
    
    for i in range(1, number + 1):
        factorial *= i
    
    print(f"Factorial of {number} is {factorial}")
```
