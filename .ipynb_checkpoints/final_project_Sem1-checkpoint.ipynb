```python
import random

# Initial wallet and crypto prices
wallet = 100000
cryptos = {"BTC": 93934.76, "ETH": 3280.10}
portfolio = {"BTC": 0, "ETH": 0}

# Function to view prices
def view_prices():
    print("\nCurrent Prices:")
    for coin, price in cryptos.items():
        print(f"{coin}: ${price:.2f}")

# Function to buy cryptocurrency
def buy_coin():
    global wallet
    coin = input("Enter the coin to buy (BTC, ETH): ").upper()
    if coin not in cryptos:
        print("Invalid coin!")
        return
    amount = float(input("Enter amount in USD to spend: "))
    if amount > wallet:
        print("Not enough money!")
        return
    portfolio[coin] += amount / cryptos[coin]
    wallet -= amount
    print(f"Bought {amount / cryptos[coin]:.6f} {coin}. Remaining wallet: ${wallet:.2f}")

# Function to sell cryptocurrency
def sell_coin():
    global wallet
    coin = input("Enter the coin to sell (BTC, ETH): ").upper()
    if coin not in portfolio or portfolio[coin] == 0:
        print("You don't own this coin!")
        return
    amount = float(input(f"Enter amount of {coin} to sell: "))
    if amount > portfolio[coin]:
        print(f"You only have {portfolio[coin]:.6f} {coin}.")
        return
    wallet += amount * cryptos[coin]
    portfolio[coin] -= amount
    print(f"Sold {amount:.6f} {coin}. New wallet balance: ${wallet:.2f}")

# Main program
def main():
    while True:
        print("\nMenu: 1. View Prices 2. Buy 3. Sell 4. Exit")
        choice = input("Choose an option: ")
        if choice == "1":
            view_prices()
        elif choice == "2":
            buy_coin()
        elif choice == "3":
            sell_coin()
        elif choice == "4":
            print("Goodbye!")
            break
        else:
            print("Invalid option!")

main()
```
