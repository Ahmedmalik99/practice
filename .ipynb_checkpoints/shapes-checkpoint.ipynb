```python
def pyramid():
    for i in range (1,6):
        for j in range(1,i+1):
            print(j , end="")
        print()
            
def staric():
    for i in range (1,6):
        for j in range(1,i+1):
            print("*" , end="")
        print()
        
def diamond():
    a=5
    for i in range(1,6):
        print(" " * (a - i) , end=" " )
        print("*" * (i*2 - 1))
    
    for j in range(5,0,-1):
        print(" " * (a-j) , end=" ")
        print("*" * (j * 2 - 1))
   
    

def number():
    for i in range(1,6):
        for j in range(1,i+1):
            print(i , end=" ")
        print()
            
def abc():
    for i in range(1,6):
        for j in range(65,65+i):
            print(chr(j) , end = "")
        print()
            


while True :
    print()
    user = int(input(" 1: pyramid \n 2: staric \n 3: diamond of staric \n 4: numbers \n 5: abc   \n 6: for exit \n Enter your desired extension: "))
    
    if user == 1:
        pyramid()
        
    elif user == 2: 
        staric()
        
    elif user == 3: 
        diamond()
        
    elif user == 4: 
        number()
        
    elif user == 5:
        abc()
        
    elif user == 6:
        print("Bye")
        break
    
    else:
        print("Please enter a number")
        
    
    
    

        
        
```
