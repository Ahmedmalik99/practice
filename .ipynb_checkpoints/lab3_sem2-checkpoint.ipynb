```python
class employee:
    def __init__(self):
        name = "Ahmed"
        role = "DevOps"
        salary = 120000
        print(f"name: {name} \nrole: {role} \nsalary: {salary}")
        
obj = employee()    
```

```python
class area:
    def __init__ (self,r):
        self.r = r
        print(2*3.142*(r**2))
        
obj = area(3)
```

```python
class A:
    def __init__(self,num):
        self.num = num
        print(f"the percentage {(num/100)*100}")        
obj = A(87)
```

```python
class BankAccount:
    def __init__(self,name,balance):
        print("welcome")
        self.n = name
        self.b = balance
    def create(self):
        if self.b <=0:
            print("not enough funds") 
        else:
            print(f"account created {self.n} with balance {self.b} ")
ob = BankAccount("Ahmed" , 120000)
ob.create()

ob = BankAccount("Ahmed" , -99)
ob.create()
```

```python
class Student:
    def __init__(self, name, age, grades):
        self.name = name
        self.age = age
        self.grades = grades
        if not grades:
            print("Student should have some grades")
        else:
            print(f"Name: {self.name}, Age: {self.age}, Grades: {self.grades}")

ob = Student("Unknown", 20, [2, 4, 5, 7])
ob2 = Student("unkown2" , 20,[])
```
