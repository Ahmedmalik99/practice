```python
# Muhammad Ahmed Malik
# 2024F-BCIS-55
import random
countries_and_capitals = {"Mexico": "Mexico City","Russia": "Moscow","South Korea": "Seoul","United Kingdom": "London","United States": "Washington, D.C","France": "Paris","Germany": "Berlin","India": "New Delhi","Italy": "Rome","Japan": "Tokyo"}
correct_count = 0
incorrect_count = 0
while True:
    country, capital = random.choice(list(countries_and_capitals.items()))
    guess = input(f"What is the capital of {country}? ").strip()
    if guess.lower() == capital.lower():
        print("Correct!")
        correct_count += 1
    elif guess.lower() == 'q':
        break
        
    else:
        print(f"Incorrect. The correct answer is {capital}.")
        incorrect_count += 1
print("\nGame over!")
print(f"Correct answers: {correct_count}")
print(f"Incorrect answers: {incorrect_count}")
```
