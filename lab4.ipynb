```python
question1 = "What is the capital of France?"
options1 = "1. Berlin\n2. Madrid\n3. Paris\n4. Rome"
correct_answer1 = 3

question2 = "What is the largest planet in our solar system?"
options2 = "1. Earth\n2. Mars\n3. Jupiter\n4. Saturn"
correct_answer2 = 3

question3 = "Who wrote 'Romeo and Juliet'?"
options3 = "1. Shakespeare\n2. Dickens\n3. Hemingway\n4. Austen"
correct_answer3 = 1

score = 0

print("Welcome to the Quiz Game!")

print(question1)
print(options1)
answer1 = int(input("Enter the number of your answer: "))
if answer1 == correct_answer1:
    print("Correct!\n")
    score += 1
else:
    print("Wrong!\n")

# Ask question 2
print(question2)
print(options2)
answer2 = int(input("Enter the number of your answer: "))
if answer2 == correct_answer2:
    print("Correct!\n")
    score += 1
else:
    print("Wrong!\n")

# Ask question 3
print(question3)
print(options3)
answer3 = int(input("Enter the number of your answer: "))
if answer3 == correct_answer3:
    print("Correct!\n")
    score += 1
else:
    print("Wrong!\n")

# Display final score
print(f"You finished the quiz! Your score is: {score}/3")
```
