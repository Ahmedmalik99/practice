```python
def sorting(arr):
    total_iterations = 0
    total_swaps = 0 
    
    print("original list: " , arr)
    
    for i in range(len(arr)):
        print(f" \n iteration number {i+1}")
        print(f" current list is {arr} \n")
        swaps_in_iteration = 0
        for j in range(len(arr) - 1 - i):
            total_iterations += 1
            print(f" camparing the number {arr[j]} and {arr[j + 1]} " , end= "")
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                total_swaps += 1 
                swaps_in_iteration =+ 1
                print(f" New list is {arr}")
            else:
                print(f"No swap")
                
        
        if swaps_in_iteration == 0:
            print("  No swaps in this iteration. The list is already sorted.")
            break
        else:
            print(f"  Total swaps in this iteration: {swaps_in_iteration}")

    print("\nSorted List:", arr)
    print("Total Iterations:", total_iterations)
    print("Total Swaps:", total_swaps)


numbers = [9,14 ,11, 8, 5]
sorting(numbers)

                
    
```
