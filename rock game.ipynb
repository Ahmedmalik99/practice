```python
import random

choices = ["rock", "paper", "scissors"]

while True:
    ans = random.choice(choices)  # Randomly pick each round
    user = input("Enter Your choice \nRock \nPaper \nScissors \nQuit\n").lower()
    
    if user == "quit":
        print("Game Over!")
        break
    elif user in choices:
        if user == ans:
            print(f"Computer = {ans} and You = {user} \nYou Won!")
        else:
            print(f"Computer = {ans} and You = {user} \nYou Lose!")
    else:
        print("Invalid input. Please choose Rock, Paper, Scissors, or Quit.")

              
```
